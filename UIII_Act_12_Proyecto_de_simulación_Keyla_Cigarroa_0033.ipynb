### Estilista

```python
import os


class Estilista:

    def __init__(
        self, id_estilista, nombre, especialidad, salon, agenda, tecnicas, marcas, premios
    ):
        self.id = id_estilista
        self.nombre = nombre
        self.especialidad = especialidad
        self.salon = salon
        self.agenda = agenda  # Ej: Lunes a Viernes
        self.tecnicas = tecnicas  # Ej: Balayage, Babylights
        self.marcas = marcas  # Ej: L'Oréal, Olaplex
        self.premios = premios  # Ej: Tijera de Oro, Ninguno

    def IA_analizar_perfil(self):
        """Simulación de IA: Sistema experto que analiza al estilista y da un consejo."""
        consejo = ""
        if "Balayage" in self.tecnicas and "Olaplex" not in self.marcas:
            consejo = "💡 IA sugiere: Deberías aliarte con 'Olaplex' para proteger el cabello en tus Balayages."
        elif int(self.premios) == 0:
            consejo = "💡 IA sugiere: ¡Inscríbete al concurso nacional de color este mes para ganar tu primer premio!"
        else:
            consejo = "💡 IA sugiere: Perfil de alta gama. Recomendado para impartir Masterclasses."
        return consejo


class SalonInteligente:

    def __init__(self):
        self.lista_estilistas = []

    def agregar_estilista(self):
        print("\n--- 💇 REGISTRAR NUEVO ESTILISTA ---")
        id_estilista = input("ID único (ej. 1, 2): ")
        nombre = input("Nombre del estilista: ")
        especialidad = input("Especialidad en color (ej. Platinados, Fantasía): ")
        salon = input("Nombre del salón de belleza: ")
        agenda = input("Agenda semanal (ej. Lun-Vie 9-17h): ")
        tecnicas = input("Técnicas dominadas (separadas por coma): ")
        marcas = input("Marcas aliadas (separadas por coma): ")
        premios = input("Cantidad de premios ganados: ")

        nuevo = Estilista(
            id_estilista, nombre, especialidad, salon, agenda, tecnicas, marcas, premios
        )
        self.lista_estilistas.append(nuevo)
        print(f"✅ ¡{nombre} ha sido registrado con éxito!")

    def listar_estilistas(self):
        print("\n" + "=" * 115)
        print(
            f"{'ID':<4} | {'NOMBRE':<15} | {'ESPECIALIDAD':<15} | {'SALÓN':<15} | {'AGENDA':<15} | {'TÉCNICAS':<20} | {'PREMIOS':<8}"
        )
        print("=" * 115)

        if not self.lista_estilistas:
            print(
                f"{'':<45} No hay estilistas registrados todavía. {'':<45}"
            )
        else:
            for e in self.lista_estilistas:
                print(
                    f"{e.id:<4} | {e.nombre:<15} | {e.especialidad:<15} | {e.salon:<15} | {e.agenda:<15} | {e.tecnicas:<20} | {e.premios:<8}"
                )
                # Aquí la IA analiza en tiempo real a cada estilista en la lista
                print(f"  {e.IA_analizar_perfil()}")
                print("-" * 115)

    def actualizar_estilista(self):
        print("\n--- 🔄 ACTUALIZAR ESTILISTA ---")
        id_buscar = input("Ingresa el ID del estilista a modificar: ")
        encontrado = False

        for e in self.lista_estilistas:
            if e.id == id_buscar:
                encontrado = True
                print(f"Modificando a: {e.nombre}")
                e.especialidad = input(
                    f"Nueva especialidad [{e.especialidad}]: "
                )
                e.salon = input(f"Nuevo salón [{e.salon}]: ")
                e.agenda = input(f"Nueva agenda [{e.agenda}]: ")
                e.tecnicas = input(f"Nuevas técnicas [{e.tecnicas}]: ")
                e.marcas = input(f"Nuevas marcas [{e.marcas}]: ")
                e.premios = input(f"Nuevos premios [{e.premios}]: ")
                print("✅ Datos actualizados correctamente.")
                break

        if not encontrado:
            print("❌ No se encontró ningún estilista con ese ID.")

    def borrar_estilista(self):
        print("\n--- 🗑️ ELIMINAR ESTILISTA ---")
        id_buscar = input("Ingresa el ID del estilista a eliminar: ")
        encontrado = False

        for e in self.lista_estilistas:
            if e.id == id_buscar:
                self.lista_estilistas.remove(e)
                print(f"🗑️ ¡{e.nombre} fue eliminado del sistema!")
                encontrado = True
                break

        if not encontrado:
            print("❌ No se encontró ningún estilista con ese ID.")


# --- MENÚ INTERACTIVO PRINCIPAL ---
def menu():
    sistema = SalonInteligente()

    # Datos demo para que los alumnos vean información desde el inicio sin escribir tanto
    estilista_demo = Estilista(
        "1",
        "Ana Gómez",
        "Rubios",
        "Glow Salon",
        "Mar-Sáb",
        "Balayage",
        "L'Oréal",
        "0",
    )
    sistema.lista_estilistas.append(estilista_demo)

    opcion = ""
    while opcion != "5":
        print("\n" + "•" * 40)
        print("🤖 SISTEMA IA: GESTIÓN DE ESTILISTAS 🤖")
        print("•" * 40)
        print("1. Agregar Estilista")
        print("2. Listar Estilistas + Análisis IA")
        print("3. Actualizar Estilista")
        print("4. Borrar Estilista")
        print("5. Salir")
        opcion = input("Selecciona una opción (1-5): ")

        if opcion == "1":
            sistema.agregar_estilista()
        elif opcion == "2":
            sistema.listar_estilistas()
        elif opcion == "3":
            sistema.actualizar_estilista()
        elif opcion == "4":
            sistema.board = sistema.borrar_estilista()
        elif opcion == "5":
            print("\n👋 ¡Gracias por usar el simulador de IA en Estilismo! Hasta pronto.")
            print("programa elaborado por Keyla Cigarroa NL = 12")
        else:
            print("⚠ Opción no válida. Intenta de nuevo.")


# Ejecutar el programa
menu()
```
